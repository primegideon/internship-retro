```python
# import the tools
import pandas as pd

# load the csv file from session storage
df = pd.read_csv('chicago_taxi_sample.csv')

# display first 5 rows to confirm load
df.head()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

# get a quick statistical summary (observe min and max for 'fare' and 'trip_miles')
display(df.describe())
```

```python
# drop any rows with missing data
df = df.dropna()

# Recheck
print("Missing values after cleaning:\n", df.isnull().sum())
```

```python
# define features (the clues)
X = df[['trip_miles', 'trip_seconds']]

# define target (what we want to predict)
Y = df['fare']

print("Features (X) size", X.shape)
print("Target (Y) size", Y.shape)
```

```python
# Import spliting tools from scikit-learn (the main python machine learning library)
from sklearn.model_selection import train_test_split

# Split clues (X) and targets (Y) into training (80%) and testing (20%) groups
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# Print results to see how many rows are in "study guide" vs "final exam"
print("Training Data (study Guide):", X_train.shape[0], "trips")
print("Testing Data (final exam):", X_test.shape[0], "trips")
```

```python
# Import the linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Create a linear regression model
model = LinearRegression()

# Train the model with the training data
model.fit(X_train, Y_train)

# Give it the final exam! algorithm predicts fares using clues from test data
predictions = model.predict(X_test)

# Predictions compared to actual real-world fares
print("The Model's Guesses: $", predictions[:5].round(2))
print("The Real Values: $", Y_test[:5].values.round(2))
```

```python
# Import the grading tool
from sklearn.metrics import mean_absolute_error

# Calculate the average dollar amount the model was off by
average_error = mean_absolute_error(Y_test, predictions)

print(f"On average, our model's fare guess is off by: ${average_error:.2f}")
```

# Create the New Clues, Retrain and Grade the Smarter Model

```python
# Convert the timestamp text into a smart date format
df['trip_start_timestamp'] = pd.to_datetime(df['trip_start_timestamp'])

# Extract the hour of the day (0 = Midnight, 23 = 11 PM)
df['trip_hour'] = df['trip_start_timestamp'].dt.hour

# Extract the day of the week (0 = Monday, 6 = Sunday)
df['trip_day'] = df['trip_start_timestamp'].dt.dayofweek

# peek at the new columns to make sure it worked
display(df[['trip_start_timestamp', 'trip_hour', 'trip_day']].head())
```

```python
# 1. Update Features (now with 4 clues!)
X_smart = df[['trip_miles', 'trip_seconds', 'trip_hour', 'trip_day']]

# 2. Split the data again so new clues are in the study guide and final exam
X_train_s, X_test_s, Y_train_s, Y_test_s = train_test_split(X_smart, Y, test_size=0.20, random_state=42)

# 3. Wake up a brand new, blank model and train it
smarter_model = LinearRegression()
smarter_model.fit(X_train_s, Y_train_s)

# 4. Take the final exam and grade it
new_predictions = smarter_model.predict(X_test_s)
new_error = mean_absolute_error(Y_test_s, new_predictions)

print("Old Average Error: $7.37")
print(f"New Average Error: ${new_error:.2f}")
```

```python
# 1. Import the Random Forest algorithm
from sklearn.ensemble import RandomForestRegressor

# 2. Wake up the smarter model (limit it to 20 trees so it runs fast)
forest_model = RandomForestRegressor(n_estimators=20, random_state=42)

# 3. Train it with the exact same study guide
forest_model.fit(X_train_s, Y_train_s)

# 4. Take the final exam and grade it!
forest_predictions = forest_model.predict(X_test_s)
forest_error = mean_absolute_error(Y_test_s, forest_predictions)

print(f"Random Forest Average Error: ${forest_error:.2f}")
```
